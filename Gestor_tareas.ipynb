```python
# =============================================
# GESTOR DE TAREAS - Proyecto de Beyfrank
# =============================================

tareas = []

def mostrar_menu():
    print("\n" + "="*40)
    print("     GESTOR DE TAREAS")
    print("="*40)
    print("1. Agregar tarea")
    print("2. Ver todas las tareas")
    print("3. Marcar tarea como completada")
    print("4. Eliminar tarea")
    print("5. Salir")
    print("="*40)

while True:
    mostrar_menu()
    opcion = input("\nElige una opción (1-5): ")

    if opcion == "1":
        tarea = input("Escribe la nueva tarea: ")
        tareas.append({"tarea": tarea, "completada": False})
        print("✅ Tarea agregada correctamente!")

    elif opcion == "2":
        print("\n📋 Tus tareas:")
        if not tareas:
            print("No tienes tareas aún.")
        else:
            for i, t in enumerate(tareas, 1):
                estado = "✅" if t["completada"] else "⭕"
                print(f"{i}. {estado} {t['tarea']}")

    elif opcion == "3":
        if not tareas:
            print("No hay tareas.")
        else:
            print("\nTareas pendientes:")
            for i, t in enumerate(tareas, 1):
                if not t["completada"]:
                    print(f"{i}. {t['tarea']}")
            try:
                num = int(input("Número de tarea a completar: ")) - 1
                tareas[num]["completada"] = True
                print("✅ Tarea marcada como completada!")
            except:
                print("Número inválido.")

    elif opcion == "4":
        if not tareas:
            print("No hay tareas.")
        else:
            for i, t in enumerate(tareas, 1):
                print(f"{i}. {t['tarea']}")
            try:
                num = int(input("Número de tarea a eliminar: ")) - 1
                eliminada = tareas.pop(num)
                print(f"🗑️ Tarea eliminada: {eliminada['tarea']}")
            except:
                print("Número inválido.")

    elif opcion == "5":
        print("👋 ¡Hasta pronto! Gracias por usar el Gestor de Tareas.")
        break

    else:
        print("❌ Opción inválida. Intenta de nuevo.")
```
